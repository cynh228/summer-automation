```python
import cv2
import numpy as np
print(cv2.__version__)  # 如果输出版本号（如 4.10.0），说明安装成功
```

```python
# 1. 读取图片
img = cv2.imread('test.jpg')  # 如果图片不在当前目录，要写完整路径

# 2. 如果图片存在，显示它
if img is not None:
    cv2.imshow('我的第一张OpenCV图片', img)
    cv2.waitKey(0)            # 等待按键，0表示无限等待
    cv2.destroyAllWindows()   # 关闭窗口
else:
    print("图片没找到，请检查路径")
```

```python
# 把刚刚读取的图片，另存为新文件
cv2.imwrite('保存的副本.jpg', img)
print("保存成功！")
```
