```python
import numpy as np
import matplotlib.pyplot as plt

# 生成 0 到 2π 之间的 100 个点（平滑曲线）
x = np.linspace(0, 2 * np.pi, 100)
# 计算对应的正弦值
y = np.sin(x)
# 创建画布和坐标轴（一句话搞定）
plt.plot(x, y, label='sin(x)', color='blue', linestyle='--', linewidth=2)

# 添加装饰
plt.title('正弦波曲线', fontsize=14)
plt.xlabel('x (弧度)')
plt.ylabel('sin(x)')
plt.legend()  # 显示图例
plt.grid(True, linestyle=':', alpha=0.6)  # 网格线

# 显示
plt.show()
```

```python
# 随机生成一些点（带噪声的正弦波）
x_scatter = np.random.rand(50) * 2 * np.pi
y_scatter = np.sin(x_scatter) + np.random.randn(50) * 0.2  # 加一些随机噪声

plt.scatter(x_scatter, y_scatter, color='red', alpha=0.6, label='带噪声的采样点')
plt.plot(x, y, color='blue', label='理想正弦波', linewidth=1)  # 叠上理想曲线
plt.title('正弦波与散点采样')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 创建一个 2行1列 的子图布局
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

# 第一个子图：正弦波
axes[0].plot(x, np.sin(x), 'b-', label='sin(x)')
axes[0].set_title('正弦波')
axes[0].legend()
axes[0].grid(True)

# 第二个子图：余弦波
axes[1].plot(x, np.cos(x), 'r-', label='cos(x)')
axes[1].set_title('余弦波')
axes[1].legend()
axes[1].grid(True)

# 调整子图之间的间距，防止重叠
plt.tight_layout()
plt.show()
```

## Matplotlib 常用操作速查

| 组件 | 函数 | 作用 |
| :--- | :--- | :--- |
| 画布和轴 | `plt.subplots()` | 创建画布和多个子图 |
| 线图 | `plt.plot(x, y)` | 绘制折线或曲线 |
| 散点图 | `plt.scatter(x, y)` | 绘制散点图 |
| 标题 | `plt.title('文字')` | 设置图表标题 |
| x轴标签 | `plt.xlabel('文字')` | 设置 x 轴说明 |
| y轴标签 | `plt.ylabel('文字')` | 设置 y 轴说明 |
| 图例 | `plt.legend()` | 显示曲线标签 |
| 网格 | `plt.grid(True)` | 显示网格线 |
| 保存图片 | `plt.savefig('文件名.png')` | 保存图表到本地 |
plt.plot 的第三个参数，写 '颜色+线型' 的速记符。
